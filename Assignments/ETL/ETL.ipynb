```python
import pandas as pd

df = pd.read_csv("Sales_data.csv")
print(df)
```

```python
df = df.dropna()
```

```python
df["tax_price"] = df["price"] * 1.18
print(df)
```

```python
import sqlite3

conn = sqlite3.connect("sales.db")
df.to_sql("sales_data", conn, if_exists="replace", index=False)

print("Data loaded into SQL database successfully")
```
