```python
#Histogram
import matplotlib.pyplot as plt
import numpy as np
data=np.random.random_sample(5)
plt.hist(data,color='green',edgecolor='yellow',bins=5)
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram plot example')
plt.show()
```

```python
#Stackplot
quaters=['q1','q2','q3','q4']
electronics=[1000,12000,11000,10500]
clothing=[5000,7500,8000,4500]
home=[3000,7500,8200,9000]
plt.figure(figsize=(10,6))
plt.stackplot(quaters,electronics,clothing,home,labels=['Electronics','Clothing','Home_appliances'])
plt.xlabel('Quaters')
plt.ylabel('Sales')
plt.title('Product Categoru Sales over QUaters')
plt.legend()
plt.show()
```

```python
#3Dplot
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
x,y=np.meshgrid(x,y)
z=np.sin(np.sqrt(x**2 + y**2))
fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')
ax.scatter(x,y,z, c=z,cmap='viridis')
plt.show()
         
```

```python
x=np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)
y3=np.tan(x)
y4=np.exp(-x)

fig,axs=plt.subplots(2,2,figsize=(10,8))
axs[0,0].plot(x,y1,'r')
axs[0,0].set_title('sine wave')
axs[0,1].plot(x,y2,'b')
axs[0,1].set_title('Cosine Wave')
axs[1,0].plot(x,y3,'g')
axs[1,0].set_title('Tangent Wave')
axs[1,1].plot(x,y4,'m')
axs[1,1].set_title('Exposed Decay')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
df=pd.read_csv(r"C:/Users/Administrator/Desktop/UST/platform_data.csv")
df
```

```python
import matplotlib.pyplot as plt

# Ranking (Higher is better → descending)
df["ROI_score"] = df["Marketing - Marketing ROI"].rank(ascending=False)
df["Conv_score"] = df["Marketing - Conversion Rate"].rank(ascending=False)
df["Enroll_score"] = df["Marketing - Student Enrolled"].rank(ascending=False)

df["CPL_score"] = df["Marketing - Cost per lead"].rank(ascending=True)

# Convert ranks to points (higher rank → more points)
max_rank = len(df)

df["Total Score"] = (
    (max_rank - df["ROI_score"] + 1) +
    (max_rank - df["Conv_score"] + 1) +
    (max_rank - df["Enroll_score"] + 1) +
    (max_rank - df["CPL_score"] + 1)
)

# Sort by total score
df = df.sort_values("Total Score", ascending=False)

# Print results
print(df[["Platform Name", "Total Score"]])

# Best company
print("\nBest Company:")
print(df.iloc[0]["Platform Name"])

plt.figure()
plt.pie(df["ROI_score"], labels=df["Platform Name"], autopct='%1.1f%%')
plt.title("ROI Performance")
plt.show()
```

```python
import matplotlib.pyplot as plt

df = df.drop_duplicates(subset="Platform Name").copy()

df["ROI_score"] = df["Marketing - Marketing ROI"].rank(ascending=False)
df["Conv_score"] = df["Marketing - Conversion Rate"].rank(ascending=False)
df["Enroll_score"] = df["Marketing - Student Enrolled"].rank(ascending=False)
df["CPL_score"] = df["Marketing - Cost per lead"].rank(ascending=True)

max_rank = len(df)

df["Total Score"] = (
    (max_rank - df["ROI_score"] + 1) +
    (max_rank - df["Conv_score"] + 1) +
    (max_rank - df["Enroll_score"] + 1) +
    (max_rank - df["CPL_score"] + 1)
)

df = df.sort_values("Total Score", ascending=False)

plt.bar(df["Platform Name"], df["Total Score"])
plt.xticks(rotation=30, ha='right')

plt.figure()
plt.pie(df["Total Score"], labels=df["Platform Name"], autopct='%1.1f%%')
plt.title("ROI Performance")
plt.show()
```
