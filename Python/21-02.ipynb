```python
class Mynumber:
    def __iter__(self):
        self.num=1
        return self
    def __next__(self):
        if self.num<=5:
            x=self.num
            self.num+=1
            return x
        else:
            raise StopIteration
myclass=Mynumber()
myiter=iter(myclass)
for x in myiter:
    print(x)
```

```python
#Genrator
def my_gen():
    for i in range(1,6):
        yield i
gen=my_gen()
for value in gen:
    print(value)
print(gen)
```
