```python
#Create Risk Score (Custom Analysis)
df['risk_score'] = (df['chol'] / 200) + (df['trestbps'] / 120) + df['oldpeak']
```

```python
def classify_risk(score):
    if score < 3: 
        return "Low Risk"
    elif score < 5:
        return "Medium Risk"
    else: return "High Risk"
        
df['risk_category'] = df['risk_score'].apply(classify_risk)
```

```python
df['risk_category'].value_counts().plot(kind='bar', color=['green','orange','red']) 
plt.ylabel("Number of Patients") 
plt.title("Risk Category Distribution") 
plt.show()
```
