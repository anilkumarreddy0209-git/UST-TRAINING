```python
#Matplot
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
moths=['Jan','Feb','Mar','Apr','May']
product_a=[45,35,40,70,80]
product_b=[10,20,30,40,50]
plt.plot(moths,product_a,marker='o',linestyle='-',color='blue',label='Prodct A')
plt.plot(moths,product_b,marker='s',linestyle='--',color='red',label='Prodct B')
plt.xlabel('Month')
plt.ylabel('Mothly Revenue')
plt.title('Monthly revenue Comparision')
plt.legend()
plt.show()
```
